# **Gradiente Descendente e Regularização**: uma abordagem matemática e computacional aplicada

O gradiente descendente e a regularização são técnicas frequentemente utilizadas em aprendizado de máquina para atingir um equilíbrio entre ajustar bem os dados de treinamento e evitar _overfitting_.

O Gradiente Descendente é um algoritmo de otimização iterativo que ajusta os parâmetros de um modelo para minimizar uma função de custo. A regularização L2 adiciona um termo de penalidade à função de custo, que é proporcional à soma dos valores quadrados dos parâmetros do modelo. Essa penalidade incentiva o modelo a ter valores de parâmetros menores, tornando-o menos sensível ao ruído nos dados de treinamento e melhorando sua capacidade de generalização para novos dados.

## **Gradiente Descendente**

O Gradiente Descendente (_Gradient Descent_) é um dos algoritmos fundamentais em otimização e aprendizado de máquina, é flexível e eficiente.

É normalmente utilizado para encontrar o mínimo de uma função objetivo, ajustando iterativamente os parâmetros do modelo na direção do gradiente negativo da função de "custo" ou **erro**.

Num modelo de regressão linear, por exemplo, o objetivo é minimizar o "risco empírico", que mede o erro entre as previsões do modelo e os valores reais.

Pode ser aprimorado com técnicas como mini-batch, regularização e métodos adaptativos, como veremos adiante.

**Aplicações do Gradiente Descendente**

O algoritmo é amplamente utilizado em várias modelagens, tais como:
- **Regressão Linear**: visando a minimização do erro quadrático médio.
- **Regressão Logística**: para otimização da função de entropia cruzada.
- **Máquinas de Vetores de Suporte (SVMs)**: Otimização do classificador de margem máxima.
- **Redes Neurais**: no treinamento de redes profundas, utilizando _backpropagation_, para ajustar os pesos da rede neural minimizando o erro entre a saída prevista e a esperada.

## **Regularização**

A regularização é uma técnica essencial em aprendizado de máquina para construir modelos mais robustos, generalizáveis e menos propensos a _overfitting_. Desempenha um papel crucial na prevenção de _overfitting_, especialmente quando os dados de treinamento limitados ou ruidosos.  Ao escolher o tipo de regularização e ajustar o hiperparâmetro **lambda**, é possível controlar a complexidade do modelo de modo a encontrar um equilíbrio ideal entre o ajuste aos dados de treinamento e a capacidade de generalização.

A combinação do Gradiente Descendente e da Regularização L2 é comumente usada em vários modelos de aprendizado de máquina, incluindo regressão linear, regressão logística e redes neurais.

A compreensão de tais técnicas é fundamental para o desenvolvimento de modelos mais eficazes e melhor ajustados, ajudando a melhorar o desempenho e a estabilidade, especialmente ao lidarmos com dados de alta dimensão ou conjuntos de dados ruidosos.


## **1. Gradiente Descendente: definição**

O objetivo do gradiente descendente é minimizar uma função $ f(\theta) $, onde $\theta$ representa os parâmetros do modelo. A cada iteração, $\theta$ é atualizado seguindo a regra:

$$
\theta^{(t+1)} = \theta^{(t)} - \alpha \nabla f(\theta^{(t)})
$$

Onde:
* $\nabla f(\theta)$ é o gradiente da função objetivo $f(\theta)$, ou seja, o vetor de derivadas parciais.
* $\alpha$ é a **taxa de aprendizado** (*learning rate*), hiperparâmetro que controla o tamanho do passo na direção negativa do gradiente.

A intuição por trás dessa abordagem é simples: a derivada indica a inclinação da função. Se a derivada é positiva, movemos $\theta$ para a esquerda (reduzindo $f(\theta)$); se for negativa, movemos $\theta$ para a direita.

## **2. Casos uni e multidimensionais**
No caso unidimensional, temos apenas um parâmetro $\theta$, e a atualização ocorre de acordo com:

$$ \theta = \theta - \alpha \frac{d}{d\theta} f(\theta) $$

Se $f(\theta)$ for uma parábola, como $f(\theta) = \theta^2$, então:

$$
\frac{d}{d\theta} f(\theta) = 2\theta
$$

A atualização se torna:

$$ \theta = \theta - \alpha (2\theta) = (1 - 2\alpha) \theta $$

No caso multidimensional, generalizamos a abordagem para um vetor de parâmetros $\theta \in \mathbb{R}^d$:

$$
\theta = \theta - \alpha \nabla f(\theta)
$$

onde o gradiente $\nabla f(\theta)$ é um vetor com as derivadas parciais de $f$ em relação a cada coordenada de $\theta$.

## **3. Escolha da Taxa de Aprendizado**
A escolha do parâmetro $\alpha$ é crucial para o desempenho do algoritmo:
- **$\alpha$ muito grande**: pode fazer o algoritmo "pular" o mínimo e divergir.
- **$\alpha$ muito pequeno**: a convergência se torna muito lenta.

Em funções convexas bem comportadas, é possível escolher um $\alpha$ pequeno o suficiente para garantir que o gradiente descendente sempre desça até o mínimo global.

## **4. Gradiente Descendente e Regularização**

O Gradiente Descendente pode ser combinado com **termos de regularização** para evitar sobreajuste. Em particular:
- **Regularização L2 (Ridge Regression)** adiciona um termo $\lambda ||\theta||^2$ na função de custo.
- **Regularização L1 (Lasso Regression)** adiciona um termo $\lambda ||\theta||_1$, incentivando sparsidade nos coeficientes.

A atualização do gradiente passa a incluir esse termo extra:

$$ \theta = \theta - \alpha (\nabla f(\theta) + \lambda \theta) $$

## **5. Convergência e Problemas Comuns**

Embora o gradiente descendente seja um método poderoso, entretanto, pode apresentar problemas:
- **Ótimos locais**: em funções não convexas, o algoritmo pode ficar preso em mínimos locais.
- **Platos e _Saddle Points_**: regiões onde o gradiente é próximo de zero podem atrasar a convergência.
- **Exploding Gradients**: o gradiente pode crescer muito rapidamente, causando instabilidade.
- **Vanishing Gradients**: em redes neurais profundas, os gradientes podem se tornar muito pequenos, dificultando o aprendizado.

Para mitigar esses problemas, é recomendado utilizar:
- **Momentum**: termo que mantém o histórico dos gradientes para evitar oscilações.
- **Adam Optimizer**: combina Momentum e RMSProp para ajustar a taxa de aprendizado adaptativamente.



Temos uma função  $f(θ) = θ₁² + θ₂²$, onde $θ$ é um vetor de duas dimensões $[θ₁, θ₂]$. Precisamos encontrar o gradiente dessa função, que é um vetor que aponta na direção do maior aumento da função.

**Cálculo das Derivadas Parciais**

O gradiente é composto pelas derivadas parciais da função em relação a cada variável.

* Derivada parcial em relação a θ₁: $∂f/∂θ₁ = 2θ₁$
* Derivada parcial em relação a θ₂: $∂f/∂θ₂ = 2θ₂$

**Montando o Vetor Gradiente**

O vetor gradiente é formado pelas derivadas parciais:

$$∇f(θ) = [∂f/∂θ₁, ∂f/∂θ₂] = [2θ₁, 2θ₂]$$



```python
# @title Plotagem de um gráfico de contorno 3D de f(θ) = θ₁² + θ₂²'
# Define a função f(θ) = θ₁² + θ₂²
def f(theta1, theta2):
  return theta1**2 + theta2**2

# Cria uma grade de valores para theta
theta1 = np.linspace(-5, 5, 100)
theta2 = np.linspace(-5, 5, 100)
theta1, theta2 = np.meshgrid(theta1, theta2)

# Calcula os valores f(θ) correspondentes
z = f(theta1, theta2)

# Cria o gráfico de superfície
# Plota a função f(θ) como uma superfície no plano (θ₂),
# considerando o plano horizontal (usual) como (θ₁) e o eixo vertical como f(θ).
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')
surf = ax.plot_surface(theta1, theta2, z, cmap=cm.viridis)
ax.set_xlabel('θ₁')
ax.set_ylabel('θ₂')
ax.set_zlabel('f(theta)')
ax.set_title('Gráfico de superfície de f(θ) = θ₁² + θ₂²')
fig.colorbar(surf)
plt.show()

# Cria o gráfico de contorno
# Plota as curvas de nível θ₁² + θ₂² = K, onde K>0 (número real fixo).
plt.figure(figsize=(8, 6))
contour = plt.contour(theta1, theta2, z, levels=20, cmap=cm.viridis)
plt.clabel(contour, inline=True, fontsize=8)
plt.xlabel('θ₁')
plt.ylabel('θ₂')
plt.title('Gráfico de contorno de f(θ) = θ₁² + θ₂²')
plt.grid(True)
plt.show()
```

## **6. Interpretação Geométrica**

Imagine a função de custo como na superfície em um espaço multidimensional acima. O gradiente descendente é como uma "bola" rolando "ladeira abaixo" nessa superfície, buscando o ponto mais baixo, que corresponderá ao mínimo da função. A taxa de aprendizado controla a velocidade da bola.

Desta forma, a taxa de aprendizado é um parâmetro crucial que precisa ser ajustado cuidadosamente. Uma **taxa muito pequena** pode levar a uma **convergência lenta**, enquanto uma **taxa muito grande** pode fazer o **algoritmo divergir**.

Entretanto, o método do gradiente pode convergir para um mínimo local, que não necessariamente é o mínimo global da função.


O método do gradiente é uma ferramenta poderosa para otimizar funções e encontrar os parâmetros que minimizam o erro em modelos de regressão linear. Ao entender os conceitos matemáticos por trás do algoritmo, podemos aplicá-lo de forma eficiente e interpretar seus resultados.

```python
# @title Cálculo do vetor do gradiente ∇_θ f(θ) para f(θ) = θ₁² + θ₂², onde θ tem 2 dimensões.
def gradient_f(theta):
    """
    Calcula o vetor gradiente da função f(θ) = θ_1^2 + θ_2^2.

    Args:
    theta (numpy.ndarray): Um array NumPy de 2 dimensões representando o vetor θ (theta).

    Returns:
    numpy.ndarray: O vetor gradiente ∇_θ f(θ) como um array NumPy.
    """
    theta_1 = theta[0]
    theta_2 = theta[1]

    # Derivada parcial em relação a θ_1 é 2 * θ_1
    df_dtheta1 = 2 * theta_1

    # Derivada parcial em relação a θ_2 é 2 * θ_2
    df_dtheta2 = 2 * theta_2

    # O gradiente é o vetor com as derivadas parciais
    gradient = np.array([df_dtheta1, df_dtheta2])

    return gradient

# Exemplo de uso:
theta = np.array([3, 4])  # Define o vetor θ
grad = gradient_f(theta) # Calcula o gradiente
print(f"O gradiente de f(θ) em θ = {theta} é: {grad}")
```

## **7. Variantes do Gradiente Descendente**

Existem variantes do gradiente descendente, como o  batch e o mini-batch gradient descent e gradiente descendente estocástico (SGD), que podem ser mais eficientes em termos computacionais, as quais exploraremos a seguir.

### **Gradiente Descendente em Lote (Batch Gradient Descent)**
Neste método, utilizamos **todos os exemplos** do conjunto de treinamento para calcular o gradiente antes de atualizar $\theta$:

$$
\theta = \theta - \alpha \frac{1}{n} \sum_{i=1}^{n} \nabla f_i(\theta)
$$

Vantagem:
- Convergência mais estável, pois usamos toda a informação disponível.

Desvantagem:
- Pode ser computacionalmente caro em grandes conjuntos de dados.



### **Mini-batch Gradient Descent**
Compromisso entre os dois anteriores: utilizamos **um subconjunto aleatório de exemplos** (mini-batch) para calcular o gradiente:

$$
\theta = \theta - \alpha \frac{1}{m} \sum_{i \in \text{mini-batch}} \nabla f_i(\theta)
$$

Vantagem:
- Mais eficiente que o batch gradient descent e menos ruidoso que o SGD.

### **Gradiente Descendente Estocástico (Stochastic Gradient Descent - SGD)**

Aqui, o gradiente é calculado **para apenas um exemplo** aleatório do conjunto de treinamento a cada iteração:

$$
\theta = \theta - \alpha \nabla f_i(\theta)
$$

Vantagem:
- Atualizações rápidas, úteis para grandes volumes de dados.

Desvantagem:
- Pode oscilar e não convergir exatamente ao mínimo global.

O Gradiente Descendente Estocástico (SGD) geralmente requer menos recursos computacionais por etapa do que a Gradiente Descendente. Isto ocorre porque a SGD atualiza os parâmetros com base no gradiente calculado a partir de um único ponto de dados ou de um pequeno lote de pontos de dados, enquanto o Gradiente Descendente **usa todo o conjunto de dados** para calcular o gradiente. Ou seja, cada etapa na SGD envolve menos cálculos, sendo portanto mais rápida.

## **8. Regularização**

A **regularização** é uma técnica essencial em aprendizado de máquina para evitar *overfitting*, especialmente quando o modelo é complexo ou os dados são ruidosos. Pois adiciona uma penalidade à função de custo para restringir os valores dos parâmetros do modelo ($\theta$), promovendo simplicidade e melhor generalização. Esse termo de penalidade "desencoraja" o modelo de ter parâmetros (pesos) muito grandes ou complexos, forçando-o a encontrar uma solução mais simples e robusta.

**Tipos de Regularização**

* **Regularização L1** (Lasso): adiciona a soma dos valores absolutos dos parâmetros à função de custo.  Isso tende a forçar alguns parâmetros a zero, efetivamente realizando a seleção de recursos.

* **Regularização L2** (Ridge): adiciona a soma dos quadrados dos parâmetros à função de custo, o que   tende a diminuir o valor dos parâmetros, tornando o modelo mais suave e menos sensível a ruídos.

### **Regularização L2 (Ridge Regression)**

A forma mais comum de regularização em problemas de regressão é a **regularização L2**, que adiciona um termo proporcional à norma L2 (quadrado) dos parâmetros à função de custo:

$$
J_{n,\lambda}(\theta, \theta_0) = \underbrace{\frac{1}{n} \sum_{i=1}^n \frac{(y^{(i)} - \theta \cdot x^{(i)} - \theta_0)^2}{2}}_{\text{Erro Quadrático Médio}} + \underbrace{\frac{\lambda}{2} ||\theta||^2}_{\text{Penalidade L2}}.
$$

- **$\lambda$ (Parâmetro de Regularização):** Controla a força da penalização.  
  - $\lambda \to 0$: O modelo se aproxima da regressão linear comum (sem regularização).  
  - $\lambda \to \infty$: Os parâmetros $\theta$ são forçados a zero, resultando em um modelo constante.

### **Efeitos da Regularização**
1. **Redução da Variância:** pois penaliza coeficientes grandes, evitando ajustes excessivos a ruídos nos dados.  
2. **Tratamento de Multicolinearidade:** torna a matriz $X^TX + \lambda I$ invertível, mesmo quando $X^TX$ é singular.  
3. **Balanceamento entre Viés e Variância:** Aumenta ligeiramente o viés para reduzir drasticamente a variância.

**Hiperparâmetro Lambda**

O hiperparâmetro lambda controla a força da regularização.  Um valor alto de lambda aumenta a penalidade, forçando o modelo a ser mais simples, enquanto um valor baixo de lambda permite que o modelo seja mais complexo.


### **Implementação com Gradiente Descendente**

#### **Atualização dos Parâmetros**
O gradiente da função de custo regularizada inclui um termo adicional $\lambda \theta$:

$$
\nabla_\theta J = \frac{1}{n} \sum_{i=1}^n \left( \theta \cdot x^{(i)} + \theta_0 - y^{(i)} \right) x^{(i)} + \lambda \theta.
$$

A regra de atualização no gradiente descendente é:

$$
\theta \leftarrow \theta - \eta \left( \nabla_\theta J \right),
$$

onde $\eta$ é a taxa de aprendizado.

### **Conclusão**
A regularização L2 é uma técnica poderosa para equilibrar viés e variância. Através do ajuste de $\lambda$, é possível controlar a complexidade do modelo e melhorar sua capacidade de generalização. O código a seguir ilustra como implementar a regressão ridge com gradiente descendente e visualizar o impacto da regularização nos coeficientes.

```python
# @title Implementação da Regularização Ridge L2 com Gradiente Descendente
# Gerar dados sintéticos
np.random.seed(42)
n = 100
d = 5
X = np.random.randn(n, d)
true_theta = np.array([2, -1, 0.5, 3, -2])
theta_0 = 1
y = X.dot(true_theta) + theta_0 + np.random.normal(0, 0.5, n)

# Hiperparâmetros
lambda_vals = [0, 0.1, 1, 10]  # Testar diferentes valores de lambda
eta = 0.01
epochs = 1000

# Função de custo com regularização L2
def compute_cost(X, y, theta, theta_0, lambda_):
    residuals = y - (X.dot(theta) + theta_0)
    mse = np.mean(residuals ** 2) / 2
    regularization = (lambda_ / 2) * np.sum(theta ** 2)
    return mse + regularization

# Gradiente Descendente
def ridge_regression(X, y, lambda_, eta, epochs):
    n, d = X.shape
    theta = np.zeros(d)
    theta_0 = 0
    cost_history = []

    for _ in range(epochs):
        residuals = y - (X.dot(theta) + theta_0)
        grad_theta = (-X.T.dot(residuals) / n) + lambda_ * theta
        grad_theta0 = -np.mean(residuals)

        theta -= eta * grad_theta
        theta_0 -= eta * grad_theta0

        cost = compute_cost(X, y, theta, theta_0, lambda_)
        cost_history.append(cost)

    return theta, theta_0, cost_history

# Treinar modelos com diferentes lambdas
results = {}
for lambda_ in lambda_vals:
    theta, theta_0, costs = ridge_regression(X, y, lambda_, eta, epochs)
    results[lambda_] = (theta, theta_0, costs)

# Visualizar coeficientes
plt.figure(figsize=(10, 6))
for lambda_, (theta, _, _) in results.items():
    plt.plot(theta, label=f'λ = {lambda_}', marker='o')

plt.xlabel('Índice do Coeficiente')
plt.ylabel('Valor do Coeficiente')
plt.title('Efeito da Regularização L2 nos Coeficientes')
plt.legend()
plt.grid(True)
plt.show()
```

**O gráfico mostra a redução dos coeficientes com o aumento do parâmetro de regularização (λ)$$

**Análise**:
- `λ = 0`: coeficientes próximos aos valores verdadeiros (sem regularização).

- `λ = 0.1` e `λ = 1`: coeficientes reduzidos em magnitude, mas ainda próximos dos verdadeiros.

- `λ = 10`: coeficientes próximos de zero, indicando subajuste (_underfitting_).
