# **Introdução as Redes Neurais**

As redes neurais são uma das ferramentas mais poderosas e versáteis no campo do aprendizado de máquina e da inteligência artificial.

São inspiradas no funcionamento do cérebro humano, onde neurônios interconectados processam informações. Ou seja, são formadas por "neurônios artificiais" que processam entradas, aplicam funções de ativação e produzem saídas.

A representação matricial e o uso de algoritmos como o gradiente descendente permitem que as redes neurais sejam treinadas para realizar tarefas complexas.

Com o avanço do deep learning, as redes neurais se tornaram uma ferramenta essencial em muitas áreas da ciência e tecnologia, sendo usadas numa variedade de aplicações, como por exemplo:

1. **Visão Computacional**: reconhecimento de objetos, detecção de faces, segmentação de imagens.

2. **Processamento de Linguagem Natural**: tradução automática, geração de texto, análise de sentimentos.

3. **Sistemas de Recomendação**: recomendação de produtos, filmes, música.

4. **Jogos**: desenvolvimento de agentes inteligentes para jogos complexos.

A seguir iremos explorar os conceitos básicos e aprofundar o nosso entendimento.

## **Conceito Básico sobre uma Rede Neural**

Uma rede neural é um modelo matemático composto por unidades de processamento chamadas neurônios artificiais (ou nós), que são organizados em camadas. Esses neurônios recebem entradas, processam-nas e produzem uma saída.

A rede neural "aprende" ajustando os pesos das conexões entre os neurônios, de modo a minimizar o erro entre as previsões e os valores reais.

No exemplo mais simples, uma rede neural pode ser representada como:

* Entrada 1 (X₁) → Peso (W₁₁) → Saída (Y)
* Entrada 2 (X₂) → Peso (W₂₁) → Saída (Y)

Onde:
* $X_1$ e $X_2$ são as entradas, $W_{11}$ e $W_{21}$ são os pesos associados a cada entrada, e $Y$ é a saída.

## **Funcionamento Matemático de um Neurônio**

A saída de um neurônio é calculada como uma combinação linear das entradas, seguida por uma função de ativação não linear. Matematicamente, expressamos como:

$$ Y = f(W_{11}X_1 + W_{21}X_2) $$

Onde:

* $f$ é a **função de ativação**, que introduz não linearidade ao modelo.

## **Representação Matricial**

Para redes neurais com múltiplas entradas e saídas, é conveniente representar as entradas, pesos e saídas como matrizes. Assim, as entradas e pesos são representados como vetores:

$$
x = \begin{bmatrix} x_1 \\ x_2 \end{bmatrix}, \quad
w = \begin{bmatrix} w_{11} \\ w_{21} \end{bmatrix}
$$

A saída $Y$ pode ser calculada como o produto escalar entre os vetores de entrada e pesos, seguido pela aplicação da função de ativação:
$$
Y = f(w^T x) = f(w_{11} x_1 + w_{21} x_2)
$$

## **Aprendizado da Rede Neural**
O aprendizado em redes neurais ocorre através do ajuste dos pesos $W$ de modo a minimizar o erro entre as previsões da rede e os valores reais. Isso é feito usando um algoritmo chamado _backpropagation_ (retropropagação), que calcula o gradiente do erro em relação aos pesos e os atualiza usando um método de otimização, como o gradiente descendente.

**Gradiente Descendente**
O gradiente descendente é um algoritmo de otimização que ajusta os pesos na direção que reduz o erro. A atualização dos pesos é dada por:

$$
W_{novo} = W_{antigo} - \eta \frac{\partial E}{\partial W}
$$
Onde:

* $η$ é a taxa de aprendizado, que controla o tamanho do passo na atualização dos pesos.

* $ \frac{\partial E}{\partial W} $ é o gradiente do erro
$E$ em relação aos pesos $W$.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# @title Rede Neural com Função de Ativação ReLU (Rectified Linear Unit), $ f(z) = \max(0, z) $
def neural_network(inputs, weights):
  """
  Aplica uma operação de rede neural com 2 entradas e 1 saída.

  Args:
    inputs: Um array NumPy com 2 elementos representando as entradas.
    weights: Um array NumPy com 2 elementos representando os pesos.

  Returns:
    A saída da rede neural após aplicar uma função de ativação (ReLU).
  """
  # calcula o produto escalar entre entradas e pesos
  weighted_sum = np.dot(inputs, weights)
  # Aplica a função de ativação ReLU
  output = np.maximum(0, weighted_sum)
  return output
```

```python
inputs = np.array([1, 2]) # Exemplo de entradas
weights = np.array([3, 4]) # Exemplo de pesos

output = neural_network(inputs, weights)

print(f"Entradas: {inputs}")
print(f"Pesos: {weights}")
print(f"Saída (z): {output}")
```

```python
# @title Rede Neural com Função de Ativação Tangente Hiperbólica (tanh), $ f(z) = \tanh(z) $
def neural_network(inputs, weights):
    """
    Executa uma operação de rede neural com 2 entradas e 1 saída, usando ativação tanh.

    Argumentos:
    inputs (np.array): Um array NumPy de forma (2, 1) representando as entradas.
    weights (np.array): Um array NumPy de forma (2, 1) representando os pesos.

    Retorna:
    np.array: Um array NumPy de forma (1, 1) representando a saída da rede neural.
    """
    z = np.dot(weights.T, inputs)
    output = np.tanh(z)
    return output
```

```python
inputs = np.array([1, 2]) # Exemplo de entradas
weights = np.array([3, 4]) # Exemplo de pesos

output = neural_network(inputs, weights)

print(f"Entradas: {inputs}")
print(f"Pesos: {weights}")
print(f"Saída (Y): {output}")

# Visualização
plt.figure(figsize=(6,4))
plt.bar(['Entrada X1', 'Entrada X2'], inputs, color='skyblue', label='Entradas')
plt.bar(['Peso W11', 'Peso W21'], weights, color='coral', label='Pesos')
plt.ylabel('Valores')
plt.title('Visualização de entradas e pesos')
plt.legend()
plt.show()
```

## **Funções de ativação**

---



A função de ativação é crucial para permitir que a rede neural aprenda padrões complexos. São funções de ativação comuns:

* ReLU (Rectified Linear Unit): $ f(x) = \max(0, x) $

* A **tangente hiperbólica** (tanh) é uma escolha comum para funções de ativação em redes neurais devido à sua natureza não linear e ao fato de ser derivável, o que é útil para algoritmos de treinamento como a retropropagação.

$$ f(x) = \tanh(x) $$

* Sigmoide: $ f(x) = \frac{1}{1+e^{-x}} $

```python
def neural_network(inputs, weights, activation='tanh'):
    """
    Executa uma operação de rede neural com múltiplas entradas e uma saída.

    Argumentos:
    inputs (np.array): Uma matriz NumPy de forma (n, 1) representando as entradas.
    weights (np.array): Uma matriz NumPy de forma (n, 1) representando os pesos.
    activation (str): A função de ativação a ser usada. Opções: 'tanh', 'relu', 'sigmoid'.
    O padrão é 'tanh'.

    Retorna:
    np.array: Uma matriz NumPy de forma (1, 1) representando a saída da rede neural.

    Gera:
    ValueError: Se as formas de entradas e pesos não corresponderem.
    """
    # Verifica se as dimensões das entradas e pesos são compatíveis
    if inputs.shape != weights.shape:
        raise ValueError("As dimensões de inputs e weights devem ser iguais.")

    # Calcula a combinação linear das entradas e pesos
    z = np.dot(weights.T, inputs)

    # Aplica a função de ativação
    if activation == 'tanh':
        output = np.tanh(z)
    elif activation == 'relu':
        output = np.maximum(0, z)
    elif activation == 'sigmoid':
        output = 1 / (1 + np.exp(-z))
    else:
        raise ValueError("Função de ativação não suportada. Use 'tanh', 'relu' ou 'sigmoid'.")

    return output

# Exemplo de uso
if __name__ == "__main__":

    inputs = np.array([1, 2]) # Exemplo de entradas
    weights = np.array([3, 4]) # Exemplo de pesos
    # inputs = np.array([[0.5], [0.3]])  # Exemplo de entradas
    # weights = np.array([[0.8], [0.2]])  # Exemplo de pesos

    # Aplica a função de ativação padrão (tanh)
    output = neural_network(inputs, weights)
    print(f"Saída com ativação tanh: {output}")

    # Usa a função de ativação ReLU
    output_relu = neural_network(inputs, weights, activation='relu')
    print(f"Saída com ativação ReLU: {output_relu}")

    # Usa a função de ativação sigmoide
    output_sigmoid = neural_network(inputs, weights, activation='sigmoid')
    print(f"Saída com ativação sigmoide: {output_sigmoid}")
```

## **Backpropagation**

O Backpropagation (Retropropagação) é um algoritmo fundamental para o treinamento de redes neurais artificiais, permite ajustar os pesos da rede neural minimizando o erro entre a saída prevista e a saída esperada, utilizando a Regra da Cadeia do cálculo diferencial para calcular os gradientes de forma eficiente.

```python
# Função de ativação sigmoide e sua derivada
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def sigmoid_deriv(z):
    return sigmoid(z) * (1 - sigmoid(z))

# Dados de entrada (X) e saída esperada (y)
X = np.array([[0,0], [0,1], [1,0], [1,1]])
y = np.array([[0], [1], [1], [0]])  # XOR

# Inicialização dos pesos e vieses
np.random.seed(42)
W1 = np.random.randn(2, 2)
b1 = np.zeros((1, 2))
W2 = np.random.randn(2, 1)
b2 = np.zeros((1, 1))
alpha = 0.1  # Taxa de aprendizado

# Treinamento da rede neural usando Backpropagation
for epoch in range(10000):
    # Forward pass
    z1 = np.dot(X, W1) + b1
    a1 = sigmoid(z1)
    z2 = np.dot(a1, W2) + b2
    a2 = sigmoid(z2)

    # Cálculo do erro
    error = a2 - y

    # Backpropagation
    delta2 = error * sigmoid_deriv(z2)
    delta1 = np.dot(delta2, W2.T) * sigmoid_deriv(z1)

    # Atualização dos pesos
    W2 -= alpha * np.dot(a1.T, delta2)
    b2 -= alpha * np.sum(delta2, axis=0, keepdims=True)
    W1 -= alpha * np.dot(X.T, delta1)
    b1 -= alpha * np.sum(delta1, axis=0, keepdims=True)

# Testando a rede treinada
print("Saída final:")
print(a2)
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def sigmoid_deriv(z):
    return sigmoid(z) * (1 - sigmoid(z))

# Generate z values
z = np.linspace(-10, 10, 100)

# Calculate sigmoid and its derivative
sigmoid_values = sigmoid(z)
sigmoid_deriv_values = sigmoid_deriv(z)

# Plotting
plt.figure(figsize=(10, 6))

plt.subplot(2, 1, 1)
plt.plot(z, sigmoid_values, label='Sigmoid')
plt.xlabel('z')
plt.ylabel('sigmoid(z)')
plt.title('Sigmoid Function')
plt.legend()
plt.grid(True)


plt.subplot(2, 1, 2)
plt.plot(z, sigmoid_deriv_values, label='Sigmoid Derivative')
plt.xlabel('z')
plt.ylabel('sigmoid_deriv(z)')
plt.title('Sigmoid Derivative Function')
plt.legend()
plt.grid(True)


plt.tight_layout()
plt.show()
```
